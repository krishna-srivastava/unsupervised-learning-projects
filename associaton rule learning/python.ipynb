```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

df = pd.read_csv("market basket data.csv")
print(df.head())
```

```python
df.drop("customer_id", axis=1,inplace=True)
```

```python
frequent = apriori(df, min_support=0.2, use_colnames=True)
print(frequent)
```

```python
rules = association_rules(
    frequent,
    metric="lift",
    min_threshold=1
)

print(rules[['antecedents',
             'consequents',
             'support',
             'confidence',
             'lift']])
```

```python
rules = association_rules(
    frequent,
    metric="confidence",
    min_threshold=0.5
)

rules = rules[rules['lift'] > 1]

print(rules[['antecedents',
             'consequents',
             'support',
             'confidence',
             'lift']])
```
